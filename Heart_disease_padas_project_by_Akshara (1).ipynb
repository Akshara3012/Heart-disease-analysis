```python
7.Create BMI Category. Classify BMI. Count each category.
```

```python
import pandas as pd
df = pd.read_csv("heart_disease_dataset.csv")
def bmi_category(x):
    if x < 18.5:
        return "Underweight"
    elif x < 25:
        return "Normal"
    elif x < 30:
        return "Overweight"
    else:
        return "Obese"
df["BMI_Category"] = df["BMI"].apply(bmi_category)
print(df["BMI_Category"].value_counts())
```

```python
import pandas as pd
df = pd.read_csv("heart_disease_dataset.csv")
import matplotlib.pyplot as plt

df["Heart_Disease"].value_counts().plot(kind="bar")
plt.title("Heart Disease Count")
plt.show()

df["Age"].plot(kind="hist", bins=10)
plt.title("Age Distribution")
plt.show()

plt.scatter(df["BMI"], df["Cholesterol_Total"])
plt.xlabel("BMI")
plt.ylabel("Cholesterol")
plt.show()

df.boxplot(column="Blood_Sugar_Fasting",
           by="Heart_Disease")
plt.show()
```
